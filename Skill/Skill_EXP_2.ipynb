```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files
uploaded = files.upload()
```

```python
df = pd.read_csv("/content/loan_sanction_test.csv");
plt.figure(figsize=(6, 4))

# Creating countplot for Loan Property Area or Education
sns.countplot(data=df, x="Property_Area", palette="coolwarm")

plt.title("Loan Applicants Count by Property Area")
plt.xlabel("Property Area")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(6, 4))

# Plotting distribution of Applicant Income
plt.hist(df["ApplicantIncome"].dropna(),
         bins=15,
         color="skyblue",
         edgecolor="black")

plt.title("Applicant Income Distribution")
plt.xlabel("Applicant Income")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(6, 5))

# Select numerical columns only
numeric_df = df.select_dtypes(include="number")

# Compute correlation matrix
corr_matrix = numeric_df.corr()

# Plot heatmap
sns.heatmap(corr_matrix,
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("Correlation Heatmap")

plt.show()
```
